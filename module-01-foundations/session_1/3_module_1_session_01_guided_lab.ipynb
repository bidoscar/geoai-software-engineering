## 3. Create a project directory

Choose a suitable working location and run:

```powershell
mkdir geoai-training
cd geoai-training
pwd
```

`mkdir` creates the folder. It does **not** move you into it; `cd` performs that second step.

## 7. Stage the file deliberately

First inspect status again:

```powershell
git status
```

Then stage:

```powershell
git add hello.py
git status
```

### Key idea
Git does not assume that every edit belongs in the next snapshot. `git add` lets you deliberately choose what will be committed.

## 8. Make the first commit

Run:

```powershell
git commit -m "Add initial Python environment check"
git log --oneline
```

If Git asks for your identity, configure the name/email you want associated with your commits, then retry the commit.

## 10. Add a .gitignore

Create `.gitignore` in VS Code with:

```text
__pycache__/
*.pyc
.venv/
.env
.ipynb_checkpoints/
```

Then:

```powershell
git status
git add .gitignore
git commit -m "Add Python gitignore"
```

**Important:** never put real secrets or API keys into Git history.

## 11. Check the branch

Run:

```powershell
git branch
```

If the current branch is `master`, rename it:

```powershell
git branch -M main
```

Verify again:

```powershell
git branch
```

## 12. Check for an existing remote

Run:

```powershell
git remote -v
```

If nothing is printed, there is no configured remote yet.

### Concept check
Git is the local version-control system. GitHub is a remote hosting/collaboration service. They are related, but they are not the same thing.

## 13. Connect to GitHub and push

Create a **new empty repository** on GitHub. Because your local repository already has history, do not initialise the remote with a README, licence or .gitignore.

Copy its repository URL, then run:

```powershell
git remote add origin <YOUR_REPOSITORY_URL>
git remote -v
git push -u origin main
```

Authentication may open a browser or use your configured Git credential method.

After the first successful upstream push, later updates can normally use:

```powershell
git push
```

## 14. Final verification

Run:

```powershell
git status
git log --oneline
git remote -v
git branch
```

A good final state is:
- working tree clean;
- meaningful commits visible;
- branch is `main`;
- `origin` points to the intended GitHub repository;
- GitHub shows the same committed files.
