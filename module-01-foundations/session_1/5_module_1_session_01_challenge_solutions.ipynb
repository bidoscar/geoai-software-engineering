## Task 4 — Example two-commit history

```powershell
git add README.md observation.py
git commit -m "Add initial environment observer"

git add .gitignore
git commit -m "Add Python gitignore"

git log --oneline
```

The exact messages may differ; the important point is that each commit represents a coherent change.

## Task 5 — Three-state model

After editing `observation.py`:

```powershell
git status
git diff
```

At this point the change is in the **working tree**.

Then:

```powershell
git add observation.py
git status
```

The selected change is now in the **staging area**.

After:

```powershell
git commit -m "Improve observation output"
```

the staged version becomes part of **repository history**.

## Task 6 — Branch and remote

```powershell
git branch -M main
git remote -v
git remote add origin <repository-url>
git remote -v
git push -u origin main
```

If `origin` already exists, do not add it again; inspect it first. The `-u` establishes the upstream relationship for the current branch.

## Task 7 — Final audit

```powershell
git status
git branch
git log --oneline
git remote -v
```

Expected reasoning: GeoAI projects evolve through code, data-processing logic, configuration and experiments. Version control makes changes traceable, reversible, reviewable and reproducible, while a remote repository supports backup and collaboration.
