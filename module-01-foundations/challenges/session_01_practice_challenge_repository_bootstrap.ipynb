# Practice Challenge 01 — Repository Bootstrap
**Module 1, Session 01**

This is formative practice, **not** Engineering Challenge 1. The Week 1 assessed challenge remains Session 6: **Environmental Data Cleaner**.


## Scenario
Establish a clean, reproducible development repository and publish its history to a remote repository.

## Required structure
```text
geoai-software-engineering/
├── hello_geoai.py
├── .gitignore
├── module-01-foundations/
│   ├── lectures/
│   ├── labs/
│   ├── exercises/
│   └── challenges/
└── learning-journal/
    └── session-01.md
```


## Requirements
1. Create the structure primarily from the terminal.
2. Execute `hello_geoai.py`.
3. Initialise Git.
4. Make multiple meaningful commits.
5. Configure `.gitignore`.
6. Keep credentials/secrets out of Git.
7. Use `main` as the primary branch.
8. Connect an empty GitHub repository as `origin`.
9. Push local history.
10. Finish with a clean working tree.


## Constraints
- Attempt independently before requesting a full solution.
- Inspect repository state before staging.
- Stage deliberately; do not use `git add .` mechanically.
- Use coherent commit messages.
- Read errors before changing commands.


## Required evidence
Record:
```powershell
git status
git log --oneline
git remote -v
git branch -vv
```
Verify the files and history are visible on GitHub.


## Technical reflection
1. What does `git init` change?
2. Why does Git have a staging area?
3. Commit vs push?
4. What is `origin`?
5. What is `origin/main`?
6. Why finish with a clean working tree?
7. Describe one failure you encountered and what it taught you.


## Completion standard
You can independently navigate the terminal, create the workspace, run Python, initialise/inspect Git, construct coherent history, configure a remote, push to GitHub, and explain the workflow.
